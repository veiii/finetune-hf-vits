```python
# Replace 'input.wav' with your filepath
input_path = "./base_audio_data/7RqQoWix3L.wav"
audio = AudioSegment.from_wav(input_path)
print(f"Loaded audio: {len(audio) / 1000:.2f} seconds, {audio.frame_rate} Hz, {audio.channels} channel(s)")
```

```python
n = 26  # seconds to skip
start_ms = n * 1000

trimmed = audio[start_ms:]  # keep everything after the first n seconds
print(f"Trimmed audio length: {len(trimmed) / 1000:.2f} seconds")
```

```python
Audio(
    data=trimmed.raw_data,
    rate=trimmed.frame_rate,
    channels=trimmed.channels
)
```

```python
buf = BytesIO()
trimmed.export(buf, format="wav")
buf.seek(0)

Audio(
    data=buf.read(),
    rate=trimmed.frame_rate
)
```

```python
trimmed.export("output.mp3", format="mp3", bitrate="192k")
```
